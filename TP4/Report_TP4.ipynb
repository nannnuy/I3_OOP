1. Encapsulation
- Summary: Wrapping data and methods into a single unit (a class). It protects internal data from direct external modification.

```python
class Car:

    def __init__(self):
        self.__max_speed = 200  # Protected via name-mangling

    def get_max_speed(self):
        return self.__max_speed  # Controlled access via method


car = Car()  #[cite: 1]
print(car.get_max_speed())  # Output: 200[cite: 1]
```

2. Abstraction
- Summary: Hiding complex implementation details. It exposes only the essential features to the user.

```python
from abc import ABC, abstractmethod  #


class Shape(ABC):  # Abstract base class

    @abstractmethod
    def area(self):  # Defines the essential interface without implementation
        pass  #[cite: 2]
```

3. Inheritance
- Summary: Allowing a child class to inherit properties and methods from a parent class. This promotes code reuse.

```python
class Animal:  # Parent class[cite: 2]

    def speak(self):
        print("Animal speaks")  #[cite: 2]


class Dog(Animal):  # Child class inheriting from Animal[cite: 2]

    def speak(self):
        print("Dog barks")  #[cite: 2]


dog1 = Dog()  #[cite: 2]
dog1.speak()  # Output: Dog barks[cite: 2]
```

4. Polymorphism
- Summary: Allowing the exact same method name to behave differently depending on which object calls it.

```python
class Dog:

    def speak(self):
        print("Dog barks")  #[cite: 2]


class Cat:

    def speak(self):
        print("Cat meows")  #[cite: 2]


animals = [Dog(), Cat()]  #[cite: 2]

for animal in animals:
    animal.speak()  # Executes the specific behavior for each object[cite: 2]
```
